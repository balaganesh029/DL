```python

!pip install -q tensorflow numpy

import tensorflow as tf
import numpy as np

print("TensorFlow version:", tf.__version__)


url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"

try:
    file_path = tf.keras.utils.get_file(
        "shakespeare.txt",
        url
    )

    with open(file_path, "r", encoding="utf-8") as f:
        text = f.read().lower()

except Exception as e:
    print("Dataset download failed.")
    print("Error:", e)
    raise

print("Dataset loaded successfully.")
print("Total characters:", len(text))


chars = sorted(set(text))

c2i = {c: i for i, c in enumerate(chars)}
i2c = {i: c for i, c in enumerate(chars)}

print("Number of unique characters:", len(chars))


seq_len = 40

X = []
y = []

max_samples = min(len(text) - seq_len, 50000)

for i in range(max_samples):

    input_seq = text[i:i + seq_len]
    target_char = text[i + seq_len]

    X.append([c2i[c] for c in input_seq])
    y.append(c2i[target_char])

X = np.array(X, dtype=np.int32)
y = np.array(y, dtype=np.int32)

print("Input shape:", X.shape)
print("Target shape:", y.shape)


model = tf.keras.Sequential([
    tf.keras.layers.Embedding(
        input_dim=len(chars),
        output_dim=64,
        input_length=seq_len
    ),

    tf.keras.layers.LSTM(128),

    tf.keras.layers.Dense(
        len(chars),
        activation="softmax"
    )
])

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam"
)

model.summary()


print("\nTraining started...\n")

history = model.fit(
    X,
    y,
    batch_size=128,
    epochs=1,
    verbose=1
)

print("\nTraining completed successfully!")

def generate(seed, length=300):

    seed = seed.lower()

    seed = "".join(c for c in seed if c in c2i)

    if len(seed) == 0:
        seed = "shall i compare thee"

    generated_text = seed

    seq = [c2i[c] for c in seed]

    for _ in range(length):

        current_seq = seq[-seq_len:]

        if len(current_seq) < seq_len:
            current_seq = (
                [0] * (seq_len - len(current_seq))
                + current_seq
            )

        inp = np.array(
            current_seq,
            dtype=np.int32
        ).reshape(1, seq_len)

        prediction = model.predict(
            inp,
            verbose=0
        )[0]


        next_idx = np.random.choice(
            len(prediction),
            p=prediction
        )

        seq.append(next_idx)

        generated_text += i2c[next_idx]

    return generated_text



seed_text = "shall i compare thee to a summer's day?"

print("GENERATED TEXT")


result = generate(
    seed_text,
    length=300
)

print(result)
```
